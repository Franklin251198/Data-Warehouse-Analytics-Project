# Raw Ingestion


```python
import pandas as pd
import os
from sqlalchemy import create engine
```

```python
engine = create_engine('sqlite:///yelp.db')
```

```python
folder_path = r'path_to_your_folder'

for file in os.listdir(folder_path):

    if file.endswith('.json'):

        try:
            file_path = os.path.join(folder_path, file)

            print(f'Loading {file}...')

            first_chunk = True

            # Read file in chunks
            for chunk in pd.read_json(file_path, lines=True, chunksize=10000):

                # Convert all columns to string
                for col in chunk.columns:
                    chunk[col] = chunk[col].astype(str)

                # Replace table if first chunk, append afterwards
                if_exists_option = 'replace' if first_chunk else 'append'

                chunk.to_sql(
                    file[:-5],
                    engine,
                    index=False,
                    if_exists=if_exists_option,
                    chunksize=1000
                )

                first_chunk = False

            print(f'{file} completed\n')

        except Exception as e:
            print(f'Error in {file}: {e}')
```
